```python
import numpy as np

# Datos del problema
valores = np.array([0, 1, 2, 3, 4, 5])   # Unidades vendidas
probabilidades = np.array([0.05, 0.10, 0.25, 0.30, 0.20, 0.10])  # Probabilidades

# Verificación
suma_prob = np.sum(probabilidades)
print("Suma de probabilidades =", round(suma_prob, 4))
```

```python
import pandas as pd

# Crear tabla de distribución
tabla = pd.DataFrame({
    "Unidades vendidas (X)": valores,
    "Probabilidad P(X)": probabilidades
})

print(tabla)
```

```python
import numpy as np
import pandas as pd

# Datos del problema
valores = np.array([0, 1, 2, 3, 4, 5])
probabilidades = np.array([0.05, 0.10, 0.25, 0.30, 0.20, 0.10])

# Distribución acumulada
F_x = np.cumsum(probabilidades)

# Crear tabla con acumulada
tabla = pd.DataFrame({
    "Unidades vendidas (X)": valores,
    "Probabilidad P(X)": probabilidades,
    "F(x) acumulada": F_x
})

print(tabla)

```

```python
import matplotlib.pyplot as plt

# Gráfica escalonada
plt.step(valores, F_x, where="post", label="F(x)")
plt.xlabel("X (Unidades vendidas)")
plt.ylabel("F(x)")
plt.title("Función de distribución acumulada")
plt.grid(True)
plt.legend()
plt.show()
```

Valor esperado (E[X]) ≈ 2.85 unidades → En promedio, se espera vender unas 3 unidades por semana.

Varianza y desviación estándar ≈ 1.5 y 1.22 → La dispersión es moderada, lo que significa que las ventas suelen estar cerca del promedio, pero pueden variar entre semanas.

Conclusión: La demanda es relativamente estable, ya que la desviación estándar no es muy alta respecto al promedio.
