```python
import numpy as np

# Datos del problema
valores = np.array([0, 10, 20, 50])   # Monto de los premios
probabilidades = np.array([0.50, 0.30, 0.15, 0.05])  # Probabilidades

# Verificación
suma_prob = np.sum(probabilidades)
print("Suma de probabilidades =", suma_prob)
```

```python
import pandas as pd

# Crear tabla de distribución
tabla = pd.DataFrame({
    "Premio (X)": ["A", "B", "C", "D"],
    "Monto (pesos)": valores,
    "Probabilidad P(X)": probabilidades
})

print(tabla)
```

```python
# Valor esperado E[X]
valor_esperado = np.sum(valores * probabilidades)

# Varianza Var(X)
varianza = np.sum(((valores - valor_esperado)**2) * probabilidades)

# Desviación estándar
desviacion = np.sqrt(varianza)

print("Valor esperado E[X] =", valor_esperado)
print("Varianza =", varianza)
print("Desviación estándar =", desviacion)
```

```python
import matplotlib.pyplot as plt

# Distribución acumulada
F_x = np.cumsum(probabilidades)

# Mostrar tabla acumulada
tabla["F(x) acumulada"] = F_x
print(tabla)

# Gráfica escalonada
plt.step(valores, F_x, where="post", label="F(x)")
plt.xlabel("X (Monto en pesos)")
plt.ylabel("F(x)")
plt.title("Función de distribución acumulada")
plt.grid(True)
plt.legend()
plt.show()
```

Valor esperado (E[X]): Representa la ganancia promedio esperada por boleto.

Si el valor esperado es positivo y mayor al costo del boleto, el juego es favorable.

Si es menor o igual al costo, el juego es desfavorable para el jugador.

En este caso, el valor esperado es 6.5 pesos.
Si el boleto cuesta más de 6.5 pesos, el juego no es favorable.
Si cuesta menos de 6.5 pesos, entonces sí conviene jugar.
