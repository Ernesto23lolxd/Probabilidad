```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Valores de X: número de células anormales
X = np.array([0, 1, 2, 3, 4])

# Probabilidades de cada valor
P = np.array([0.10, 0.20, 0.35, 0.25, 0.10])

# Verificar que las probabilidades sumen 1
suma = np.sum(P)

print("Suma de las probabilidades:", suma)

if np.isclose(suma, 1):
    print("Las probabilidades suman 1 correctamente.")
else:
    print("Error: las probabilidades no suman 1.")
```

```python


# Crear la tabla de distribución
tabla = pd.DataFrame({
    "Número de células (X)": X,
    "Probabilidad P(X)": P
})

display(tabla)

```

```python


# Calcular el valor esperado
media = np.sum(X * P)

# Calcular E(X^2)
media_cuadrado = np.sum((X**2) * P)

# Calcular la varianza
varianza = media_cuadrado - media**2

# Calcular la desviación estándar
desviacion = np.sqrt(varianza)

print("Valor esperado E(X):", round(media, 4))
print("Varianza Var(X):", round(varianza, 4))
print("Desviación estándar:", round(desviacion, 4))

```

```python


# Calcular las probabilidades acumuladas
F = np.cumsum(P)

# Crear la tabla acumulada
tabla_acumulada = pd.DataFrame({
    "X": X,
    "P(X)": P,
    "F(X) = P(X <= x)": F
})

display(tabla_acumulada)
```

```python


plt.figure(figsize=(8, 5))

plt.bar(X, P, edgecolor="black")

plt.xlabel("Número de células anormales (X)")
plt.ylabel("Probabilidad P(X)")
plt.title("Distribución de probabilidad")
plt.xticks(X)
plt.grid(axis="y", alpha=0.3)

plt.show()

```

```python
print("INTERPRETACIÓN BIOMÉDICA")
print(f"Promedio esperado: {media:.2f} células anormales por muestra.")
print(f"Varianza: {varianza:.4f} células^2.")
print(f"Desviación estándar: {desviacion:.4f} células.")

print(f"\nProbabilidad de encontrar como máximo 1 célula: {F[1]*100:.1f}%")
print(f"Probabilidad de encontrar como máximo 2 células: {F[2]*100:.1f}%")
print(f"Probabilidad de encontrar como máximo 3 células: {F[3]*100:.1f}%")
print(f"Probabilidad de encontrar 4 células: {P[4]*100:.1f}%")
```

Valor esperado (E[X]) ≈ 2.35 células anormales → En promedio, se espera encontrar alrededor de 2 a 3 células anormales por muestra.

Varianza ≈ 1.47 y desviación estándar ≈ 1.21 → La variabilidad es moderada, lo que significa que la mayoría de las muestras estarán cerca del promedio, aunque algunas pueden mostrar más o menos células.

Función acumulada:

Permite ver la probabilidad de encontrar un número igual o menor de células anormales.

Por ejemplo, la probabilidad de encontrar máximo 2 células anormales es la suma de P(0), P(1) y P(2) ≈ 0.65 (65%).

Esto indica que es más probable encontrar pocas células anormales que muchas.
