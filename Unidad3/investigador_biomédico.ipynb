```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Valores de X: número de células anormales
X = np.array([0, 1, 2, 3, 4])

# Probabilidades de cada valor
P = np.array([0.10, 0.20, 0.35, 0.25, 0.10])

# Verificar que las probabilidades sumen 1
suma = np.sum(P)

print("Suma de las probabilidades:", suma)

if np.isclose(suma, 1):
    print("Las probabilidades suman 1 correctamente.")
else:
    print("Error: las probabilidades no suman 1.")
```

```python

tabla = pd.DataFrame({
    "X": X,
    "P(X)": P,
    "X * P(X)": X * P,
    "X^2 * P(X)": X**2 * P
})

display(tabla)
```

```python

# Valor esperado
media = np.sum(X * P)

# Valor esperado de X al cuadrado
media_cuadrado = np.sum((X**2) * P)

# Varianza
varianza = media_cuadrado - media**2

# Desviación estándar
desviacion = np.sqrt(varianza)

print("Valor esperado E(X):", round(media, 4))
print("Varianza Var(X):", round(varianza, 4))
print("Desviación estándar:", round(desviacion, 4))
```

```python

# Calcular la probabilidad acumulada
F = np.cumsum(P)

# Crear tabla acumulada
tabla_acumulada = pd.DataFrame({
    "X": X,
    "P(X)": P,
    "F(X) = P(X <= x)": F
})

display(tabla_acumulada)
```

```python

plt.figure(figsize=(8, 5))

plt.bar(X, P, edgecolor="black")

plt.xlabel("Número de células anormales (X)")
plt.ylabel("Probabilidad P(X)")
plt.title("Distribución de probabilidad")
plt.xticks(X)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.step(X, F, where="post", linewidth=2)
plt.scatter(X, F)

plt.xlabel("Número de células anormales (x)")
plt.ylabel("F(x)")
plt.title("Función de distribución acumulada")
plt.xticks(X)
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)

plt.show()
```

```python

print("INTERPRETACIÓN BIOMÉDICA")

print(f"El promedio esperado es de {media:.2f} células anormales por muestra.")

print(f"La desviación estándar es de {desviacion:.2f} células.")

print("\nProbabilidades acumuladas:")

for x, f in zip(X, F):
    print(f"La probabilidad de encontrar como máximo {x} "
          f"células anormales es de {f*100:.1f}%.")
```
