Es una distribución de Poisson.

¿Por qué?

El problema ya indica que:

"El número xxx de personas ingresadas a una unidad de cuidados intensivos en un hospital particular, en un día, tiene una distribución de probabilidad de Poisson con media igual a cinco personas por día."

Además, las características coinciden con una distribución de Poisson:

Se cuenta el número de ocurrencias (ingresos a UCI).
Los eventos ocurren en un intervalo de tiempo fijo (un día).
Los ingresos son independientes entre sí.
Se conoce la tasa promedio de ocurrencia: λ=5\lambda = 5λ=5 personas por día.

Por lo tanto:

X∼Poisson(λ=5)X \sim Poisson(\lambda=5)X∼Poisson(λ=5)

```python
# Problema 1 - Inciso A
from scipy.stats import poisson

# Parámetros
media_ingresos = 5
x = 2

# Probabilidad de exactamente 2 ingresos
probabilidad = poisson.pmf(x, media_ingresos)

print("Media (λ):", media_ingresos)
print("Número de ingresos:", x)
print("Probabilidad de exactamente 2 ingresos =", round(probabilidad, 6))
print("Porcentaje =", round(probabilidad * 100, 2), "%")
```

```python
# Problema 1 - Inciso B
from scipy.stats import poisson

# Parámetros
media_ingresos = 5

# Probabilidad de más de 10 ingresos
probabilidad_mayor_10 = 1 - poisson.cdf(10, media_ingresos)

print("Probabilidad de que los ingresos excedan de 10:")
print(round(probabilidad_mayor_10, 6))
print("Porcentaje =", round(probabilidad_mayor_10 * 100, 2), "%")
```

```python
from scipy.stats import poisson

# =================================
# DATOS DEL PROBLEMA
# =================================
media_ingresos = 5

print("DISTRIBUCIÓN UTILIZADA: POISSON")
print("λ =", media_ingresos)

# =================================
# INCISO A
# =================================
x = 2

probabilidad_2 = poisson.pmf(x, media_ingresos)

print("\nINCISO A")
print("P(X = 2) =", probabilidad_2)
print("Porcentaje =", round(probabilidad_2 * 100, 2), "%")

# =================================
# INCISO B
# =================================
probabilidad_mayor_10 = 1 - poisson.cdf(10, media_ingresos)

print("\nINCISO B")
print("P(X > 10) =", probabilidad_mayor_10)
print("Porcentaje =", round(probabilidad_mayor_10 * 100, 2), "%")

if probabilidad_mayor_10 < 0.05:
    print("\nConclusión:")
    print("NO es probable que los ingresos excedan de 10 personas.")
else:
    print("\nConclusión:")
    print("Sí es probable que los ingresos excedan de 10 personas.")
```
