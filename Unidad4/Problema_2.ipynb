Aunque el problema podría modelarse con una Binomial porque:

Hay un número fijo de motores: n=1000n=1000n=1000.
Cada motor puede ser defectuoso o no defectuoso.
La probabilidad de defecto es constante: p=0.001p=0.001p=0.001.

Se utiliza Poisson porque:

nnn es muy grande (100010001000).
ppp es muy pequeño (0.0010.0010.001).
Se cumple la aproximación de Poisson a la Binomial.

Calculamos:

λ=np=(1000)(0.001)=1\lambda=np=(1000)(0.001)=1λ=np=(1000)(0.001)=1

Por lo tanto:

X∼Poisson(λ=1)X \sim Poisson(\lambda=1)X∼Poisson(λ=1)

donde XXX representa el número de motores defectuosos en el embarque.

```python
from scipy.stats import poisson

# Parámetro de Poisson
lam = 1000 * 0.001

# Probabilidad de 0 defectuosos
p0 = poisson.pmf(0, lam)

print("Probabilidad de que ningún motor sea defectuoso:")
print(round(p0, 6))
print("Porcentaje =", round(p0*100, 2), "%")
```

```python
from scipy.stats import poisson

# ======================================
# DATOS DEL PROBLEMA
# ======================================

n = 1000
p = 0.001

# Parámetro λ de Poisson
lam = n * p

print("DISTRIBUCIÓN UTILIZADA: POISSON")
print("λ =", lam)

# ======================================
# A) Ningún defectuoso
# ======================================

p0 = poisson.pmf(0, lam)

print("\nA) Ningún motor defectuoso")
print("P(X=0) =", round(p0, 6))
print("Porcentaje =", round(p0*100, 2), "%")

# ======================================
# B) Tres defectuosos
# ======================================

p3 = poisson.pmf(3, lam)

print("\nB) Tres motores defectuosos")
print("P(X=3) =", round(p3, 6))
print("Porcentaje =", round(p3*100, 2), "%")

# ======================================
# C) Cuatro defectuosos
# ======================================

p4 = poisson.pmf(4, lam)

print("\nC) Cuatro motores defectuosos")
print("P(X=4) =", round(p4, 6))
print("Porcentaje =", round(p4*100, 2), "%")
```
