```python
#Basado en el libro estadistica para administracion pagina 12
# creamos una lista vacia que representa la edad de 100 personas
import random
edades = []
# Hacemos un iclo para generar 100 numeros aleatorios que representan la edad de las personas
for i in range(100):
  # genearmos un numero aleatorio con la funcion randint y lo guardamos en la variable
  # numero_aleatorio
  numero_aleatorio = random.randint(18,70)
  #Agregamos el numero aleatorio a la lista de valores
  edades.append(numero_aleatorio)
print(edades)
```

```python
import pandas as pd

#crear un dataframe de pandas con los numeros aleatorios
df = pd.DataFrame(edades, columns=['Edad'])
df.head()
```

```python
#calcular la tabla de distribucion de frecuencias
tabla_frecuencias = df['Edad'].value_counts().reset_index()
tabla_frecuencias.columns = ['Edad', 'Frecuencia']
tabla_frecuencias = tabla_frecuencias.sort_values(by='Edad')

#imprimir la tabla de distribucion de frecuencias
print("Tabla de distribucion de frecuencias:")
tabla_frecuencias
```

```python
# calcular la frecuencia relativa
tabla_frecuencias['Frec Relativa'] = tabla_frecuencias['Frecuencia'] / 100
tabla_frecuencias
```

```python
#Calcular la frecuencia porcentual
tabla_frecuencias['Frec Relativa'] = tabla_frecuencias['Frecuencia'] / 100 * 100
tabla_frecuencias
```

```python
#cualcular la frecuencia acumulada
tabla_frecuencias['Frecuencia Acumulada'] = tabla_frecuencias['Frecuencia'].cumsum()
tabla_frecuencias
```

```python
# Importamos la libreria math de funciones matematicas
import math

#volvemos a usar la lista edades
print(edades)
```

```python
#crear un dataframe con las edades
df = pd.DataFrame({'Edad': edades})
df
```

```python
#Calcular el numero de intervalos utilizando la regla de sturges
n = len(edades)
k = int( 1 + math.log2(n))
print("Valor de n = ", n , "\nNumeros de intervalos =", k)
```

```python
#Crear la tabla de frecuencia

tabla_frecuencias = pd.cut( df['Edad'], bins=k, include_lowest=True).value_counts().reset_index()
tabla_frecuencias.columns = ['Clase','Frec']
tabla_frecuencias
```

```python
# ordenar la tabla de frecuencia por el intervalo
tabla_frecuencias = tabla_frecuencias.sort_values(by='Clase')
tabla_frecuencias
```

```python
#Generar columnas siguientes
tabla_frecuencias['Frec Rel'] = tabla_frecuencias['Frec'] / 100
tabla_frecuencias['Frec Porc'] = tabla_frecuencias['Frec'] / n * 100
tabla_frecuencias['Frec Ac'] = tabla_frecuencias['Frec'].cumsum()
tabla_frecuencias['Frec Rel AC'] = tabla_frecuencias['Frec Rel'].cumsum()
tabla_frecuencias['Frec Porc AC'] = tabla_frecuencias['Frec Porc'].cumsum()

# Mostrar la tabla de frecuencia
print(tabla_frecuencias)
```
