```python
from statistics import *
```

```python
#Lista de calificaciones del grupo del grupo a de probalbidad y estadistica
grupoa = [70, 70, 65, 90, 100, 95, 45, 80, 73, 20, 100, 96]
print(grupoa)
type(grupoa)
```

```python
# sacar la media de las calicaciones con una fomrula
promedio = sum(grupoa) /len(grupoa)
print(promedio)
```

```python
#Utilizar la funcion mean que esta en la libreria statistics prara sacar el promedio

promedio2 = mean(grupoa)
print(promedio2)
```

```python
#determinar el valor de la mediana con la funcion median

mediana = median(grupoa)
print(mediana)
```

##Determinar la moda del ejemplo 1


```python
#Determinar el valor de la moda con la funcion mode

moda= mode(grupoa)
print(grupoa)
print(moda)
```

```python
#En caso de que exista mas de una moda, se puede emplear la funcion multimode para sacar las multiples moda que puedan existir

moda2= multimode(grupoa)
print(moda2)
```

```python
#Definimos una variable llamada tiempos100mts de tipo lista, que contiene los tiempos que tardaron los 10 alumnos para correr los 100 metos
tiempos100mts = [15.10, 17.20, 14.69, 13.27, 22.15, 18.71, 19.15, 20.65, 15.10, 17.20]

#desplegamos los valores de la lista
print(tiempos100mts)
```

```python
#Usamos la funcion median para determinar el tiempo promedio de la lista de de valores

#definimos una variable llamada mediana para calclar la mediana de la lista con la funcion
# median

mediana = median(tiempos100mts)
print("la media de los corredores es=>", mediana)
```

```python
#Definimos una varianle llamada moda para calcular la moda de la lista con la funcion
#mode
moda =mode(tiempos100mts)
print("La moda de los corredores es=>", moda)
```

```python
#Libreria para generar numeros aleatorios
import random
```

```python
fabrica1 = []
for i in range(100):
  numero_aleatorio = round(random.uniform(50,100),2)
  fabrica1.append(numero_aleatorio)
print(fabrica1)
```

```python
fabrica2 = []
for i in range(100):
  numero_aleatorio = round(random.uniform(70,100),2)
  fabrica2.append(numero_aleatorio)
print(fabrica2)
```

```python
#libreia de python para graficar valores
import matplotlib.pyplot as plt
```

```python
plt.hist(fabrica1)
plt.title("Produccion diaria de la fabrica 1")
# Text(0.5, 1.0, 'Produccion Diaria de la fabrica 1')
```

```python
plt.hist(fabrica2)
plt.title("Produccion diaria de la fabrica 2")
 # Text(0.5, 1.0, 'Produccion Diaria de la fabrica 1')
```

```python
media1= mean(fabrica1)
mediana1 = median(fabrica1)
moda1 = mode(fabrica1)

print("valores de la fabrica 1")
print("la media es =", mediana1)
print("la mediana es=", mediana1)
print("la moda es =", moda1)
```

```python
media2= mean(fabrica2)
mediana2 = median(fabrica2)
moda2 = mode(fabrica2)

print("valores de la fabrica 2")
print("la media es =", mediana2)
print("la mediana es=", mediana2)
print("la moda es =", moda2)
```
